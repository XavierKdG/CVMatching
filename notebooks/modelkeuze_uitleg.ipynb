# TSDAE voor CV ↔ Job Retrieval
In dit notebook leggen we uit welke keuzes zijn gemaakt voor het trainen van een TSDAE-model op vacaturedata en hoe we CV ↔ Job retrieval evalueren.

## 2. Keuze van technieken en modellen

### 2.1 SentenceTransformer
- **Model:** `all-MiniLM-L6-v2`
- **Waarom:** Lichtgewicht, snel en goed in semantische tekstrepresentaties; ondersteunt TSDAE training en is eenvoudig.

### 2.2 TSDAE (Transformer-based Denoising AutoEncoder)
- **Doel:** Unsupervised fine-tuning op domeinspecifieke job descriptions.
- **Waarom:** Geen gelabelde data nodig, verbetert embeddings, verwijdert ruis en vat context beter samen.

### 2.3 Cosine Similarity voor retrieval
- **Waarom:** Standaard metric voor vectorvergelijking; Top-K ranking identificeert meest relevante jobs per CV.

## 3. Data preprocessing
- **Bron:** `job_descriptions2_cleaned.csv` (kolom: Job Description)
- **Splitsing:** Train 70%, Val 15%, Test 15%
- **Waarom:** Trainen op grootste deel, testen op nooit eerder geziene data.

## 4. Training setup
- **Dataset:** `DenoisingAutoEncoderDataset`
- **Dataloader:** batch_size=64, shuffle=True
- **Loss:** `DenoisingAutoEncoderLoss` (reconstructie van noisy zinnen)
- **Optimizer:** lr=2e-5
- **Epochs:** 1 (snelle prototyping)
- **Output:** `./models/tsdae_model_last`
- **Motivatie:** Batch size 64 → balans GPU/geheugen; 1 epoch → snelle verificatie; lr 2e-5 → standaard fine-tuning.

## 5. Evaluatie: CV ↔ Job retrieval
- **Functie:** `retrieval_evaluate(model, cvs, jobs, k=5)`
- **Werking:** Encodeer CVs & jobs → cosine similarity → gemiddelde Top-K scores per CV
- **Waarom Top-K:** Simuleert praktijkscenario’s (meest relevante matches tonen)
- **Output:** Average Top-K Cosine Similarity als metric voor semantische matching
