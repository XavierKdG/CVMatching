# Samenvatting Evaluatieresultaten CV ↔ Job Matching

We vergelijken twee modellen: 
1. **all-MiniLM-L6-v2** (pretrained)
2. **TSDAE model** (fine-tuned op vacatures)

Evaluaties:
- Mutual Ranking (MRR): gemiddelde ranking van juiste matches
- Perturbation Drop: robuustheid voor belangrijke keywords
- Hubness: hoe vaak jobs in Top-K matches van CV’s voorkomen

```python
# Resultaten in Python dictionary
results = {
    'all-MiniLM-L6-v2': {
        'mean_mutual_mrr': 0.14925,
        'perturbation_drop': 0.000208,
        'mean_hubness': 6.746,
        'max_hubness': 180
    },
    'tsdae_model_last': {
        'mean_mutual_mrr': 0.15080,
        'perturbation_drop': 0.000210,
        'mean_hubness': 6.746,
        'max_hubness': 175
    }
}
```

## Observaties
- **MRR:** TSDAE model is iets beter (0.1508 vs 0.1492) → betere ranking van juiste matches.
- **Perturbation Drop:** bijna gelijk en zeer laag → beide modellen robuust.
- **Hubness:** ongeveer gelijk, TSDAE iets lager max hubness → geen grote verandering.

💡 **Interpretatie:**
- Fine-tuning met TSDAE verhoogt de kwaliteit van de matches iets.
- Trade-off: in grotere datasets zie je vaak dat hogere MRR samengaat met hogere hubness, omdat sommige jobs vaker als top-match verschijnen.
- Hier blijft hubness ongeveer gelijk, dus het model verbetert ranking zonder extra dominantie van populaire jobs.
- Ook heb ik de tsdae model getrained voor 1 epoch dus de resultaat verschilt niet zo erg van de base model
